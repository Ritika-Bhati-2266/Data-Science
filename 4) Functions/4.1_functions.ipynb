```python
def print_numbers(*krish):
    for number in krish:
        print(number)
print_numbers(1,4,7,8,9,4,6,8,"Hello")
```

```python
def print_numbers(*args):
    for num in args:
        print(num)
print_numbers(1,3,4,6,8)
```

```python
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
print_details(name = "Krish", age = 32, country = "India")
```

```python
def print_details(*args,**kwargs):
    for num in args:
        print(f" Positional argument: {num}")
    for key, value in kwargs.items():
        print(f"{key}:{value}")
print_details(1,3,2,4,6,7,"krish",name="Krish", age=32)
# print_details(1,3,2,4,6,7,"krish",name="Krish", age=32,8,9)
```
