```python
# Syntax
lambda arguments :expression 
```

```python
def add(a,b):
    return a+b
add(2,6)
```

```python
add = lambda a,b :a+b
print(add(2,6))
```

```python
def even(num):
    if num%2==0:
        return True
even(24)
```

```python
even = lambda num: num%2==0
print(even(6))
```

```python
def add(x,y,z):
    return x+y+z
add(2,5,8)
```

```python
add=lambda x,y,z:x+y+z
add(4,6,1)
```

```python
numbers=[1,2,3,4,5,6]
def square(num):
    return num**2
square(2)
```

```python
square=lambda x:x**2
square(2)
```

```python
map(lambda x:x**2,numbers)
```
